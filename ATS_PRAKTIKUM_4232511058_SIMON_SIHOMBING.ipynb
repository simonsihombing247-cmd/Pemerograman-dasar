```python
#SOAL 1: Daya generator dan efisiensi turbin
#cara kerja:
#program menyimpan parameter fisik kedalam variabel, menghitung daya
#menggunakan perkalian(*), menghitung efisiensi menggunakan operator
#pembagian(/), lalu menampilkan hasilnya dengan format f-string

tegangan = 220
arus = 15
energi_input = 500
energi_output = 400

daya_listrik = tegangan*arus
efisiensi = (energi_output/energi_input)*100

print(f"daya listrik = {daya_listrik}w")
print(f"efisiensi = {efisiensi:2f}%")
```

```python
#soal 3: data suhu boiler
#cara kerja
#program memanipulasi struktur data list menggunakan indeks(nulai dari 0)
#teknik slicing [awal:akhir], pungsi. append() untuk menambah data di akhir.
#dan fungsi len()untuk menghitung total elemen dalam list.

suhu = [92.5, 88.7, 95.1, 102.3, 97.8]
print(suhu[2])
print(suhu[:3])
suhu.append(99.4)
print(len(suhu))
```

```python
#soal 5: katup pengaman
#cara kerja
#program membaca angka input desimal (float) dari pengguna , lalu mengecek
#kondisi secara berurutan menggunakan struktur percabangan if,elif,else.
#dan operator logika(and) untuk menentukan status katup.

tekanan_uap = float(input("masukkan tekanan uap(psi):"))
status_manual_override = False

if tekanan_uap>150 and status_manual_override == False:
  print("katup terbuka")
elif status_manual_override == True:
  print("katup mode manual")
else:
  print("katup tertutup")
```

```python
#soal 7: pemanas mencapapai terget
#cara kerja
#perulangan while akan terus mengeksekusi blok kode di dalam nya selama kondisi
#(suhu_saat_ini<85) bernilai true . setiap putaran menambah suhu 5,
#mencetaknya, dan mengecek apakah suhu sampai 80 untuk memberi peringatan.

suhu_saat_ini = 60
while suhu_saat_ini < 85:
  suhu_saat_ini += 5
  print(f"suhu saat ini : {suhu_saat_ini}c")
  if suhu_saat_ini == 80:
    print("WARNING: Mendekati target")

print("target 85 c tercapai")
```

```python
#soal 9 : pembacaan vibrasi dengan sensor rusak.
#cara kerja:
# perulangan for menelusuri setiap item di dalam list 'data' beserta nomor urutnya
# menggunakan fungsi enumerate (..., start=1). blok try-except menangkap error
# konveksi teks ke float agar program tidak crash, dan break menghentikan loop.
# seketika jika ditemukan vibrasi kritis (>3.0)


data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

invalid_count = 0

for i, val in enumerate(data, start=1):
    try:
        v = float(val)
    except ValueError:
        print(f"Unit {i}: data tidak valid")
        invalid_count += 1
        continue

    if v > 3.0:
        print(f"Warning: Unit {i} vibrasi tinggi!")
        print("KRITIS: pemeriksaan dihentikan")
        break
    elif v > 2.0:
        print(f"Warning: Unit {i} vibrasi tinggi!")

print(f"Jumlah data tidak valid: {invalid_count}")
```
